```python
import os
from gensim.models import KeyedVectors

cache_dir = r"C:\Users\Lenovo\.cache\kagglehub\datasets\suraj520\googlenews-vectors-negative300bingz-gz-format\versions\1"
full_path = os.path.join(cache_dir, "GoogleNews-vectors-negative300.bin")  # no .gz

model = KeyedVectors.load_word2vec_format(full_path, binary=True, limit=500000)
```
